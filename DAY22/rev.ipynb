# BAR CHARTS

Key characteristics of a bar chart:


The x-axis represents the different categories.

The y-axis shows the frequency or value associated with each category.

### plt.bar()

```python
# Calculate counts of each penguin species
species_counts = penguins['species'].value_counts()
```

```python
# Bar chart to visualize species counts
plt.bar(species_counts.index, species_counts)
```

```python
species_counts=penguins['species'].value_counts()

plt.figure(figsize=(8,4))
plt.bar(species_counts.index,species_counts)
plt.title('Penguin species count')
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
penguins = sns.load_dataset('penguins')

# Calculate counts of each penguin species
species_counts = penguins['species'].value_counts()

# Pie chart of species counts
plt.figure(figsize=(5, 5))
plt.pie(species_counts, labels=species_counts.index)
plt.title('Penguin Species Distribution')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
penguins = sns.load_dataset('penguins')

# Calculate counts of each penguin species
species_counts = penguins['species'].value_counts()

# Pie chart of species counts with percentages
plt.figure(figsize=(5, 5))
# TODO: Include the 'autopct' parameter with a value of '%1.1f%%'
plt.pie(species_counts, labels=species_counts.index,autopct='%1.1f%%')
plt.title('Penguin Species Distribution')
plt.show()
```
