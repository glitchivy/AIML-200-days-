# LINE PLOTS

### Loading the Penguins Dataset

```python
penguins=sns.load_dataset("penguins")
```

Before we start plotting the data, it's helpful to understand the context of the penguins dataset. This dataset comes from the research conducted on penguin species in the Palmer Archipelago, Antarctica. It includes several measurements for three different species of penguins: Adelie, Chinstrap, and Gentoo.

The dataset features the following columns:

species: The species of the penguin (Adelie, Chinstrap, or Gentoo).
island: The island in the Palmer Archipelago where the penguin was observed.
bill_length_mm: The length of the penguin’s bill in millimeters.
bill_depth_mm: The depth of the penguin’s bill in millimeters.
flipper_length_mm: The length of the penguin’s flipper in millimeters.
body_mass_g: The mass of the penguin's body in grams.
sex: The sex of the penguin (male or female).

### setting up FIGSIZE

```python
plt.figure(figsize=(8,4)) #setting size of the plot
plt.plot(penguins['flipper_length_mm'], label='Flipper Length')
plt.title('Flipper Length Over Samples')  # Add a title
plt.xlabel('Sample Number')  # Add x-axis label
plt.ylabel('Flipper Length (mm)')  # Add y-axis label
plt.legend()  # Add a legend
plt.show() #display the plot
```

# Creating Scatter Plots with Matplotlib

Scatter plots are a type of graph used to display and compare two sets of numerical data. Each point on the scatter plot represents one observation from your data set.

The position of a point is determined by two variables:

One is shown on the x-axis (horizontal).
The other is shown on the y-axis (vertical).

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,4))
plt.scatter(penguins["bill_length_mm"],penguins['flipper_length_mm'])
plt.title('Bill_lenght VS flipper_lenght')
plt.xlabel("Bill lenght(mm)")
plt.ylabel("Flipper Lenght(mm)")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
penguins = sns.load_dataset('penguins')

# Scatter plot of bill length vs. flipper length
plt.figure(figsize=(8, 4))
# TODO: Insert the correct function call to create the scatter plot
plt.scatter(penguins['bill_length_mm'], penguins['flipper_length_mm'])

plt.title('Bill Length vs. Flipper Length')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Flipper Length (mm)')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
penguins = sns.load_dataset('penguins')

# Scatter plot of bill length vs. flipper length
plt.figure(figsize=(8, 4))
# TODO: Add the alpha parameter to the plot
plt.scatter(penguins['bill_length_mm'], penguins['flipper_length_mm'],alpha=0.5)
plt.title('Bill Length vs. Flipper Length')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Flipper Length (mm)')
plt.show()
```
