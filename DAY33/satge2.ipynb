```python
import numpy as np

x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])
def predict(w, b, x):
    return w * x + b

def compute_errors(y_predicted,y_actual):
    return y_predicted-y_actual

def compute_cost(w,b,x,y):
    y_predicted=predict(w,b,x)
    errors = compute_errors(y_predicted,y)
    sum_no= np.sum(errors**2)
    return sum_no/(2*len(y_predicted))


print(compute_cost(0, 0, x, y))
parameter_sets = [
    (0, 0),
    (1, 0),
    (2, 0),
    (2, 1)
]

for w, b in parameter_sets:
    print("w =", w, "b =", b)
    print("Predictions:", pre(w, x, b))
    print("Cost:", cost(w, x, b, y))
    print()
```

```python
import numpy as np

#predcit 

def pre(w,x,b):
    return w*x+b 
def err(predicted,y):
    return predicted-y 

def cost(w,x,b,y):
    predicted_value=pre(w,x,b)
    error=err(predicted_value,y)
    square=error**2
    sum_error=np.sum(square)
    divide=sum_error/(2*len(predicted_value))
    return divide
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])
print(cost(0,x,0,y))
print("Cost A:", cost(1, x, 0, y))
print("Cost B:", cost(2, x, 0, y))
print("Cost C:", cost(2, x, 1, y))



parameter_sets = [
    (0, 0),
    (1, 0),
    (2, 0),
    (2, 1)
]

for w, b in parameter_sets:
    print("w =", w, "b =", b)
    print("Predictions:", pre(w, x, b))
    print("Cost:", cost(w, x, b, y))
    print()
```

### Linear Regression Equation

y_predicted = w * x + b

### Cost Function

J(w, b) = sum((y_predicted - y)²) / (2 * m)

- A cost of zero means every prediction matches its actual value.

```python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([3.0, 5.0, 7.0, 9.0])

def predict(x,w,b):
    return w*x+b 

def compute_cost(y_predicted,y_actual):
    error= y_predicted-y_actual
    squared=error**2
    sum_error=sum(squared)
    return sum_error/(2*len(y_predicted))

w=2
b=1
y_predicted=predict(x,w,b)
print(y_predicted)
print(compute_cost(y_predicted,y))

experiments = [
    (0, 0),
    (1, 0),
    (2, 0),
    (2, 1)
]

for w,b in experiments:
    y_predicted=predict(x,w,b)
    cost=compute_cost(y_predicted,y)
    print(f"w={w}, b={b}")
    print("Predictions:", y_predicted)
    print("Cost:", cost)
    print()
```
