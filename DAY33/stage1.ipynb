## Question 3 of 8: Aggregation and Mean Squared Error (MSE)

### Objective
Practice NumPy aggregation functions, element-wise operations, and calculating the MSE used in linear regression.

### Given Data

You are given actual values and predicted values:

```python
import numpy as np

actual = np.array([10, 20, 30, 40])
predicted = np.array([12, 18, 33, 35])

Calculate the sum of all elements in actual using np.sum().

Calculate the average of actual using np.mean().

Find the maximum value in actual using np.max().

Find the minimum value in actual using np.min().

Calculate the prediction errors using actual - predicted.

Square each prediction error individually using ** 2.

Calculate the Mean Squared Error (MSE) using np.mean() on the squared errors.

````python
import numpy as np

actual = np.array([10, 20, 30, 40])
predicted = np.array([12, 18, 33, 35])

print(np.sum(actual))
print(np.mean(actual))
print(np.max(actual))
print(np.min(actual))
error=actual-predicted
print(error**2)
print(np.mean((error**2)))
````
