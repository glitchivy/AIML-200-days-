# Question 17 — Annotate a Data Point

### Task

Create a line plot and add a text annotation pointing out the **highest value**.

### Data

```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 40, 30]

````python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,25,15,40,30]
plt.plot(x,y)
y_max = max(y)
x_max = x[y.index(y_max)]
plt.annotate("Highest", xy=(x_max, y_max))
````

annotate is used to highlight the values

# Question 19 — Add Reference Lines

### Task

Create a line plot and add:
- A horizontal reference line at `y = 25`
- A vertical reference line at `x = 3`

### Data

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

````python
x=[1,2,3,4,5]
y=[10,25,15,40,30]
plt.plot(x,y,linestyle=':',color='green',linewidth=2.0,marker='o')
plt.axhline(30)
plt.axvline(3)
plt.show()
````
