# Phase 1 — Matplotlib Test

## Challenge

Create **one complete line chart** using the data below.

```python
x = [1, 2, 3, 4, 5]

sales_2025 = [20, 35, 30, 45, 50]
sales_2026 = [25, 30, 40, 50, 60]

````python
x = [1, 2, 3, 4, 5]
sales_2025 = [20, 35, 30, 45, 50]
sales_2026 = [25, 30, 40, 50, 60]
import matplotlib.pyplot as plt

plt.title("2025 VS 2026 Sales data ")
plt.plot(x,sales_2025,label='2025',color='red',marker='o',linewidth=1.5,linestyle='--',markersize=3,markerfacecolor='red')
plt.plot(x,sales_2026,label='2026',color='green',marker='o',linewidth=1.5,linestyle=':',markersize=3,markerfacecolor='green')
plt.ylabel("Sales")
plt.xlabel('Month')
plt.legend()
plt.show()
````
