# Question 21 — Two Lines on One Graph

### Task

Plot two different datasets on the **same graph**.

### Data

```python
x = [1, 2, 3, 4, 5]

sales_2025 = [20, 35, 30, 45, 50]
sales_2026 = [25, 30, 40, 50, 60]

````python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]

sales_2025 = [20, 35, 30, 45, 50]
sales_2026 = [25, 30, 40, 50, 60]
plt.title("sales report 2025 vs 2026")

plt.plot(x,sales_2025,label='line1',color='red')
plt.plot(x,sales_2026,label='line2')
plt.legend()
plt.xlabel("Year")
plt.ylabel("Sales")
plt.show()
````

# Question 23 — Create Subplots (1×2)

### Task

Create **two plots side by side** using a `1 × 2` subplot layout.

### Data

```python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]

````python
import matplotlib.pyplot as plt 
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
fig, ax = plt.subplots(1, 2)
ax[0].plot(x,sales)
ax[0].set_title("sales")
ax[1].plot(x,profit)
ax[1].set_title("profit")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[1].set_xlabel("Month")
ax[1].set_ylabel("Profit")

plt.show()
````

# Question 24 — Create Subplots (2×2)

### Task

Create **four plots in a 2 × 2 layout**.

### Data

```python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

````python
fig,ax=plt.subplots(2,2)
x = [1, 2, 3, 4, 5]
sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

ax[0][0].plot(x,sales)
ax[0][0].set_title("Sales")
ax[0][1].plot(x,profit)
ax[0][1].set_title("profit")
ax[1][0].plot(x,expenses)
ax[1][0].set_title("expenses")
ax[1][1].plot(x,customers)
ax[1][1].set_title("customers")
plt.tight_layout()

````

# Question 25 — Shared X-Axis

### Task

Create two subplots vertically, one for Sales and one for Profit.

### Data

```python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]

````python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
fig,ax=plt.subplots(2,1,sharex=True)
ax[0].plot(x,sales)
ax[1].plot(x,profit)
ax[0].set_title("sales")
ax[1].set_title("Profit")
ax[1].set_xlabel("profit")
ax[0].set_xlabel("Sales")
plt.tight_layout()
````

# Question 27 — Different Charts in Different Subplots

### Task

Create two subplots side by side, but use a **different chart type** in each.

### Data

```python
categories = ["HR", "IT", "Sales", "Finance"]

employees = [20, 45, 60, 30]

satisfaction = [3.2, 4.1, 3.8, 4.5]

````python
import matplotlib.pyplot as plt

categories = ["HR", "IT", "Sales", "Finance"]

employees = [20, 45, 60, 30]
satisfaction = [3.2, 4.1, 3.8, 4.5]

fig, ax = plt.subplots(1, 2)

# Left: Bar chart
ax[0].bar(categories, employees)
ax[0].set_title("Employees by Department")
ax[0].set_xlabel("Department")
ax[0].set_ylabel("Employees")

# Right: Line chart
ax[1].plot(categories, satisfaction, marker='o')
ax[1].set_title("Employee Satisfaction")
ax[1].set_xlabel("Department")
ax[1].set_ylabel("Satisfaction")

plt.tight_layout()
plt.show()
````

# Question 28 — Adjust Subplot Spacing

### Task

Create a `2 × 2` subplot layout and adjust the spacing between the plots.

### Data

```python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(x, sales)
ax[0, 0].set_title("Sales")

ax[0, 1].plot(x, profit)
ax[0, 1].set_title("Profit")

ax[1, 0].plot(x, expenses)
ax[1, 0].set_title("Expenses")

ax[1, 1].plot(x, customers)
ax[1, 1].set_title("Customers")

plt.subplots_adjust(wspace=0.4, hspace=0.5)

plt.show()
````

# Question 29 — Common Title for All Subplots

### Task

Create a `2 × 2` subplot layout and give the **entire figure one common title**.

### Data

```python
x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [20, 35, 30, 45, 50]
profit = [10, 20, 18, 30, 35]
expenses = [15, 18, 25, 20, 28]
customers = [50, 60, 55, 75, 90]

fig, ax = plt.subplots(2, 2)
fig.suptitle("company overalll performance")
ax[0, 0].plot(x, sales)
ax[0, 0].set_title("Sales")

ax[0, 1].plot(x, profit)
ax[0, 1].set_title("Profit")

ax[1, 0].plot(x, expenses)
ax[1, 0].set_title("Expenses")

ax[1, 1].plot(x, customers)
ax[1, 1].set_title("Customers")


plt.tight_layout()

plt.show()
````
