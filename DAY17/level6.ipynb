### DATASET 

```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department_ID": [1, 2, 2, 3, 4, 5],
    "Salary": [45000, 60000, 55000, 80000, 50000, 75000]
})


departments = pd.DataFrame({
    "Department_ID": [1, 2, 3, 4, 6],
    "Department": ["HR", "IT", "Finance", "Marketing", "Sales"],
    "City": ["Mumbai", "Delhi", "Pune", "Mumbai", "Chennai"]
})


performance = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 105, 107],
    "Performance": ["A", "B", "A", "C", "A"]
})


projects = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Project": [
        "AI Chatbot",
        "Cloud Migration",
        "Data Warehouse",
        "Fraud Detection",
        "Marketing Dashboard",
        "ERP System"
    ]
})


new_employees = pd.DataFrame({
    "Employee_ID": [107, 108],
    "Name": ["George", "Helen"],
    "Department_ID": [6, 2],
    "Salary": [65000, 70000]
})


print("Employees")
print(employees)

print("\nDepartments")
print(departments)

print("\nPerformance")
print(performance)

print("\nProjects")
print(projects)

print("\nNew Employees")
print(new_employees)
```

# Pandas Level 6 - Question 76

## Topic
**Inner Join (`merge()`)**

---

## Problem Statement

Using the existing DataFrames:

- `employees`
- `departments`

Perform an **inner join** on the `Department_ID` column.

Store the merged DataFrame in a variable named `result` and print it.

```python
result=pd.merge(employees,departments,on="Department_ID",how='inner')
print(result)
```

# Pandas Level 6 - Question 77

---

## Problem Statement

Using the existing DataFrames:

- `employees`
- `departments`

Perform a **left join** on the `Department_ID` column.

Store the merged DataFrame in a variable named `result` and print it.

---

```python
result=pd.merge(employees,departments,on='Department_ID',how='left')
print(result)
```
