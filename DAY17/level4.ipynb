# Pandas Level 4 - Question 62

## Topic
**Apply a Function to One Column (`apply()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), increase every employee's **Salary** by **10%** using the `apply()` method.

Store the updated Salary column in a variable named `result` and print it.

---

## Requirements

Write a program that:

- Uses the existing DataFrame.
- Applies a function to the `Salary` column.

# Pandas Level 4 - Question 63

## Topic
**Apply a Function to the Entire DataFrame (`applymap()`)**

> **Note:** In newer versions of Pandas, `DataFrame.map()` is replacing `applymap()`, but interview questions and older codebases still commonly use `applymap()`. We'll learn `applymap()` first because it's what you'll encounter most often.

---

## Problem Statement

Using the existing DataFrame (`df`), convert **every string value** in the DataFrame to **uppercase**.

Leave numeric columns unchanged.

Store the updated DataFrame in a variable named `result` and print it.


````python
result = df.map(lambda x: x.upper() if isinstance(x, str) else x)
print(result)
````

# Pandas Level 4 - Question 64

## Topic
**Map Values (`map()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), convert the values in the **Department** column using the following mapping:

```python
{
    "HR": "Human Resources",
    "IT": "Information Technology",
    "Finance": "Accounts",
    "Marketing": "Sales"
}
```

Store the updated column in a variable named `result` and print it.

````python
result = df["Department"].map({
    "HR": "Human Resources",
    "IT": "Information Technology",
    "Finance": "Accounts",
    "Marketing": "Sales"
})

print(result)
````
