# Pandas Level 5 - Question 72

## Topic

---

## Problem Statement

Using the existing DataFrame (`df`), find the following statistics for the **Salary** column department-wise:

# Pandas Level 5 - Question 73

## Topic
**Custom Aggregation (`agg()` with a Custom Function)**

---

## Problem Statement

Using the existing DataFrame (`df`), find the **range of Salary** for each **Department**.

The range is calculated as:

```text
Maximum Salary - Minimum Salary
```

Use a **custom aggregation function** with `agg()`.

Store the result in a variable named `result` and print it.


````python
result=df.groupby(by='Department')['Salary'].agg(lambda x:x.max()-x.min())
print(result)
````

# Pandas Level 5 - Question 74

## Topic
**Filter Groups (`groupby().filter()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), keep **only those departments** where the **average Salary is greater than 60,000**.

Use `groupby().filter()`.

Store the filtered DataFrame in a variable named `result` and print it.


````python
result=df.groupby(by='Department').filter(lambda x:x['Salary'].mean()>60000)
print(result)
````

# Pandas Level 5 - Question 75

---

## Problem Statement

Use `groupby().transform()`.

# 66. Mean

Average value of each group.

```python
result = df.groupby("Department")["Salary"].mean()
```

# 67. Count

Count non-null values in each group.

```python
result = df.groupby("Department")["Employee_ID"].count()
```

---

# 68. Max

Highest value in each group.

```python
result = df.groupby("Department")["Salary"].max()
```

---

# 69. Min

Lowest value in each group.

```python
result = df.groupby("Department")["Salary"].min()
```

---

# 70. Sum

Total value of each group.

```python
result = df.groupby("Department")["Salary"].sum()
```

---

# 71. Group By Multiple Columns

Group using more than one column.

```python
result = df.groupby(["Department", "City"])["Salary"].mean()
```

---

# 72. agg()

Perform multiple aggregations at once.

```python
result = df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min"]
)
```

Named Aggregation (Modern)

```python
result = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Highest_Salary=("Salary", "max"),
    Lowest_Salary=("Salary", "min")
)
```

---

# 73. Custom Aggregation

Create your own aggregation function.

Using a function

```python
def salary_range(x):
    return x.max() - x.min()

result = df.groupby("Department")["Salary"].agg(salary_range)
```

Using lambda

```python
result = df.groupby("Department")["Salary"].agg(
    lambda x: x.max() - x.min()
)
```

---

# 74. filter()

Keep only groups satisfying a condition.

```python
result = df.groupby("Department").filter(
    lambda x: x["Salary"].mean() > 60000
)
```

Example:

Keep departments having at least 3 employees.

```python
result = df.groupby("Department").filter(
    lambda x: len(x) >= 3
)
```

---

# 75. transform()

Returns one value for every original row.

```python
df["Department_Avg_Salary"] = (
    df.groupby("Department")["Salary"]
      .transform("mean")
)
```

Another example

```python
df["Department_Max_Salary"] = (
    df.groupby("Department")["Salary"]
      .transform("max")
)
```

# Difference Between agg(), filter() and transform()

## agg()

Returns one row per group.

```python
df.groupby("Department")["Salary"].agg(["mean", "max"])
```

Output

```
HR
IT
Finance
Marketing
```

---

## filter()

Returns only groups satisfying a condition.

```python
df.groupby("Department").filter(
    lambda x: x["Salary"].mean() > 60000
)
```

Output

```
Entire Finance group
```

---

## transform()

Returns same number of rows as original DataFrame.

```python
df.groupby("Department")["Salary"].transform("mean")
```

Output

```
48333
48333
48333
58750
58750
...
```

# Interview Questions

### Difference

```python
df["Salary"].mean()
```

vs

```python
df.groupby("Department")["Salary"].mean()
```

- First → Overall average
- Second → Average per department

---

### Difference

```python
count()
```

vs

```python
size()
```

- `count()` ignores NaN
- `size()` counts every row

---

### Difference

```python
map()
```

vs

```python
replace()
```

- `map()` → Missing keys become NaN
- `replace()` → Missing keys stay unchanged

---

### Difference

```python
agg()
```

vs

```python
transform()
```

| agg() | transform() |
|--------|-------------|
| One row per group | Same rows as original |
| Summarizes data | Creates new features |

---

### Difference

```python
filter()
```

vs

```python
agg()
```

| filter() | agg() |
|-----------|-------|
| Removes groups | Summarizes groups |
