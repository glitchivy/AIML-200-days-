# Titanic Dataset: Feature Classification

## Feature Classification Table

|  # | Feature       | Meaning                                | Classification                    | Important Note                                  |
| -: | ------------- | -------------------------------------- | --------------------------------- | ----------------------------------------------- |
|  1 | `survived`    | Whether the passenger survived         | **Binary + Categorical + Target** | `0` = No, `1` = Yes                             |
|  2 | `pclass`      | Passenger ticket class                 | **Ordinal**                       | 1st > 2nd > 3rd in class ranking                |
|  3 | `sex`         | Passenger sex                          | **Categorical + Binary**          | `male` / `female`                               |
|  4 | `age`         | Passenger age                          | **Numerical**                     | Continuous numerical variable                   |
|  5 | `sibsp`       | Number of siblings/spouses aboard      | **Discrete Numerical**            | Counts people, so values are whole numbers      |
|  6 | `parch`       | Number of parents/children aboard      | **Discrete Numerical**            | Counts people, so values are whole numbers      |
|  7 | `fare`        | Passenger ticket fare                  | **Numerical**                     | Continuous numerical variable                   |
|  8 | `embarked`    | Port where passenger boarded           | **Categorical**                   | `C`, `Q`, `S`                                   |
|  9 | `class`       | Passenger class in text form           | **Ordinal**                       | First, Second, Third                            |
| 10 | `who`         | Passenger category                     | **Categorical**                   | `man`, `woman`, `child`                         |
| 11 | `adult_male`  | Whether passenger was an adult male    | **Boolean + Binary**              | `True` / `False`                                |
| 12 | `deck`        | Passenger's deck                       | **Categorical**                   | A, B, C, etc.; contains many missing values     |
| 13 | `embark_town` | Name of boarding town                  | **Categorical**                   | Southampton, Cherbourg, Queenstown              |
| 14 | `alive`       | Whether passenger was alive            | **Binary + Categorical**          | `yes` / `no`; essentially duplicates `survived` |
| 15 | `alone`       | Whether passenger was travelling alone | **Boolean + Binary**              | `True` / `False`                                |

# Titanic EDA - Final Insights

## 1. Overall Survival

- Total passengers: **891**
- Survived: **342 (38.4%)**
- Did not survive: **549 (61.6%)**
- Majority of passengers did not survive.

---

## 2. Sex and Survival

| Sex | Survival Rate |
|---|---:|
| Female | **~74%** |
| Male | **~19%** |

### Insight
Sex showed one of the **strongest relationships with survival**.

Female passengers had a much higher survival rate than male passengers.

---

## 3. Passenger Class and Survival

| Pclass | Survival Rate |
|---|---:|
| 1st | **63.0%** |
| 2nd | **47.3%** |
| 3rd | **24.2%** |

### Insight
Survival generally decreased as passenger class moved from **1st → 3rd class**.

---

## 4. Sex + Pclass

| Pclass | Female | Male |
|---|---:|---:|
| 1st | **96.8%** | **36.9%** |
| 2nd | **92.1%** | **15.7%** |
| 3rd | **50.0%** | **13.5%** |

### Insight
Combining variables revealed a much stronger pattern than looking at either variable separately.

The combination of **sex and passenger class** was one of the strongest survival patterns found.

---

## 5. Fare and Survival

The boxplot showed that survivors generally paid **higher fares** than non-survivors.

### Insight
Fare was associated with survival.

However:

> **Association does not mean causation.**

Fare is also related to passenger class, so the relationship between fare and survival may partly reflect the strong relationship between **Pclass and survival**.

---

## 6. Age and Survival

The overall age distributions of survivors and non-survivors overlapped substantially.

- Survivor median age: approximately **27**
- Non-survivor median age: approximately **25**

### Insight
Age alone showed a **weaker relationship** with survival compared with sex and Pclass.

However, the `who` grouping showed:

| Group | Survival Rate |
|---|---:|
| Woman | **75.6%** |
| Child | **59.0%** |
| Man | **16.4%** |

This showed that grouping a continuous variable into meaningful categories can reveal patterns that are less obvious from the raw distribution.

---

## 7. Embarked and Survival

| Embarked | Survival Rate |
|---|---:|
| C | **55.4%** |
| Q | **39.0%** |
| S | **33.7%** |

### Insight

Passengers embarking at **C had the highest observed survival rate**.

However, the port itself should not be assumed to cause survival.
