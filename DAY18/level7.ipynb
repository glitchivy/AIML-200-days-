# Pandas Level 7 - Question 84

## Topic
**Create a Pivot Table (`pivot_table()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), create a pivot table where:

- **Rows** = `Department`
- **Columns** = `City`
- **Values** = `Salary`
- **Aggregation Function** = `mean`

Store the result in a variable named `result` and print it.

````python
result=df.pivot_table(index='Department',columns='City',values='Salary',aggfunc='mean')
print(result)
````

# Pandas Level 7 - Question 85

## Topic
**Reshape Data using `melt()`**

---

## Problem Statement

Using the existing DataFrame (`df`), convert the **Math** and **Science** columns into rows.

Keep the following columns unchanged:

- `Employee_ID`
- `Name`

Store the result in a variable named `result` and print it.


````python
result = df.melt(
    id_vars=["Employee_ID", "Name"],
    value_vars=["Math", "Science"],
    var_name="Subject",
    value_name="Marks"
)

print(result)
````

# What is Reshaping?

Reshaping means **changing the structure of a DataFrame** without changing the data.

Example:

Before

| Department | City | Salary |
|------------|------|--------|
| HR | Mumbai | 45000 |
| HR | Delhi | 48000 |
| IT | Mumbai | 60000 |

↓

After Pivot

| Department | Mumbai | Delhi |
|------------|--------|-------|
| HR | 45000 | 48000 |
| IT | 60000 | NaN |

---

# Pivot vs GroupBy

## GroupBy

Summarizes data.

```python
df.groupby("Department")["Salary"].mean()
```

Output

```
HR         46500
IT         60000
Finance    80000
```

---

## Pivot

Changes layout.

```python
df.pivot(
    index="Department",
    columns="City",
    values="Salary"
)
```

Output

```
            Mumbai   Delhi
HR           45000   48000
IT           60000     NaN
```

# Level 7 Overview

### pivot_table()

# Interview Questions

### pivot vs pivot_table

- pivot → No duplicates
- pivot_table → Handles duplicates using aggregation

---

### melt vs pivot

- melt → Wide → Long
- pivot → Long → Wide

---

### stack vs unstack

- stack → Columns become rows
- unstack → Rows become columns

---

### groupby vs pivot_table

- groupby → Summary
- pivot_table → Summary + Reshaping
