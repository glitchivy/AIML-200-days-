# Level 9: String Operations (94–97)

### DATASET

````python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID":[101,102,103,104,105],
    "Name":[
        "Alice Johnson",
        "Bob Smith",
        "Charlie Brown",
        "David Lee",
        "Eva Green"
    ],
    "Email":[
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@gmail.com",
        "david@outlook.com",
        "eva@gmail.com"
    ],
    "Department":[
        "Human Resources",
        "Information Technology",
        "Finance",
        "Marketing",
        "Human Resources"
    ]
})

print(df)
````

# Pandas Level 9 - Question 94

## Topic
**Convert Text to Uppercase (`.str.upper()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), convert all values in the **Department** column to uppercase.

Store the result in a variable named `result` and print it.

---

````python
result = df["Department"].str.upper()
print(result)
````

# Pandas Level 9 - Question 95

## Topic
**Split Strings (`.str.split()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), split the values in the **Email** column into two parts:

- Username
- Domain

using the `@` symbol.

Store the result in a variable named `result` and print it.

````python
result = df["Email"].str.split("@")
print(result)
````

# Pandas Level 9 - Question 96

## Topic
**Extract Text using Regular Expressions (`.str.extract()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), extract the **username** from the `Email` column.

Example:

```text
alice@gmail.com
```

↓

```text
alice
```

Use `.str.extract()`.

Store the result in a variable named `result` and print it.

---

````python
result = df["Email"].str.extract(r"(.*)@")

print(result)
````

# Pandas Level 9 - Question 97

## Topic
**Replace Text (`.str.replace()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), replace the domain **gmail.com** with **outlook.com** in the **Email** column.

Store the updated emails in a variable named `result` and print it.
