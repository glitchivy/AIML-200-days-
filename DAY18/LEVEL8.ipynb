# DATASET

````python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101,102,103,104,105,106,107,108],
    "Name": [
        "Alice","Bob","Charlie","David",
        "Eva","Frank","Grace","Helen"
    ],
    "Joining_Date": [
        "2021-01-15",
        "2020-07-22",
        "2022-03-10",
        "2019-11-05",
        "2021-08-18",
        "2023-02-27",
        "2020-12-30",
        "2022-06-14"
    ],
    "Resignation_Date": [
        "2023-01-15",
        "2024-07-22",
        "2023-09-10",
        "2022-11-05",
        "2024-08-18",
        None,
        "2023-12-30",
        None
    ],
    "Salary":[45000,60000,55000,80000,50000,75000,52000,68000]
})

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df["Resignation_Date"] = pd.to_datetime(df["Resignation_Date"])

print(df)
````

# Pandas Level 8 - Question 89

## Topic
**Extract Year (`.dt.year`)**

---

## Problem Statement

Using the existing DataFrame (`df`), create a new column named **Joining_Year** by extracting the year from the **Joining_Date** column.

Store the updated DataFrame in a variable named `result` and print it.

````python
df['joining_Year']=df["Joining_Date"].dt.year
print(df)
````


# Pandas Level 8 - Question 90

---

## Problem Statement

Using the existing DataFrame (`df`), create a new column named **Joining_Month** by extracting the month from the **Joining_Date** column.

Store the updated DataFrame in a variable named `result` and print it.

````python
df['Joining_Month']=df['Joining_Date'].dt.month
print(df)
````

# Pandas Level 8 - Question 91

## Topic
**Extract Weekday (`.dt.day_name()`)**

---

## Problem Statement

Using the existing DataFrame (`df`), create a new column named **Joining_Day** that contains the **day of the week** (e.g., Monday, Tuesday) for each employee's `Joining_Date`.

Store the updated DataFrame in a variable named `result` and print it.


````python
df['Joining_Day']=df['Joining_Date'].dt.day_name()
print(df)
````

# Pandas Level 8 - Question 92

## Topic
**Filter Rows Using Dates**

---

## Problem Statement

Using the existing DataFrame (`df`), display only those employees who joined **after January 1, 2021**.

Store the filtered DataFrame in a variable named `result` and print it.

````python
result = df[df["Joining_Date"] > "2021-01-01"]

print(result)
````

---

## Problem Statement

Using the existing DataFrame (`df`), create a new column named **Tenure_Days** that contains the number of days each employee worked.

Calculate it as:

```text
Resignation_Date - Joining_Date
```

Store the updated DataFrame in a variable named `result` and print it.
