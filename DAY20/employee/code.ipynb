# task 4

```python
workforce =[
    "Department",
    "Gender",
    "JobRole",
    "EducationField",
    "OverTime",
    "MaritalStatus",
    "Attrition"
]
```

```python
for i in workforce:
    print("\nUnique Categories:")
    print(df[i].nunique())

    print("\nValue Counts:")
    print(df[i].value_counts())

    print("\nPercentage Distribution:")
    print(df[i].value_counts(normalize=True) * 100)

    print("\nMost Common Category:")
    print(df[i].mode()[0])

    print("\nLeast Common Category:")
    print(df[i].value_counts().idxmin())
```
