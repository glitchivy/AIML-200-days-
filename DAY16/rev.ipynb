```python
import pandas as pd

# Dataset of apparel prices where there might be outliers
prices = [19.99, 25.99, 22.50, 27.99, 250.00, 23.49, 19.45]
df = pd.DataFrame(prices, columns=['Price'])

# Identifying and filtering out potential outliers
Q1, Q3 = df['Price'].quantile([0.25, 0.75])
IQR = Q3 - Q1
df = df[~((df['Price'] < (Q1 - 1.5 * IQR)) | (df['Price'] > (Q3 + 1.5 * IQR)))]
print(df)
```

### `pd.merge()`

```python
import pandas as pd

# Dataset 1: Basic information about some students
students = pd.DataFrame({
    'student_id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [15, 16, 17]
})

# Dataset 2: Additional performance information about the same students
performance = pd.DataFrame({
    'student_id': [1, 3, 4],
    'grade': ['A', 'B', 'A'],
    'attendance': [95, 85, 100]
})
```

```python
students_merge=pd.merge(students,performance,on='student_id',how='left')
print(students_merge)
```

# Different Types of Joins:

### LEFT JOIN

```python
left_join=pd.merge(students,performance,on='student_id',how='left')
print(left_join)
```

### RIGHT JOIN

```python
right_join=pd.merge(students,performance,on='student_id',how='right')
print(right_join)
```

### INNER JOIN

```python
inner_join=pd.merge(students,performance,on='student_id',how='inner')
print(inner_join)
```

### outer join

```python
outer_join=pd.merge(students,performance,on='student_id',how='outer')
print(outer_join)
```

# Grouping Data

```python
# Import pandas library
import pandas as pd

# Create the sales data as a dictionary
data = {
    'Representative': ['Alice', 'Bob', 'Alice', 'Bob', 'Charlie', 'Charlie'],
    'Region': ['East', 'West', 'West', 'East', 'East', 'West'],
    'Sales': [150, 200, 100, 250, 175, 300]
}

# Convert the dictionary to a DataFrame
df = pd.DataFrame(data)
print(df)
```

# `groupby`

```python
group=df.groupby('Representative')
print(group)
```

### Applying Functions to Groups: Summing Sales

```python
total_sales=df.groupby('Representative')['Sales'].sum()

print(total_sales)
```

### Applying Functions to Groups: Counting Entries

```python
count_sales=df.groupby('Representative')['Sales'].count()
print(count_sales)
```

### Applying Functions to Groups: Average Sales

```python
avg_sales=df.groupby('Representative')['Sales'].mean()
print(avg_sales)
```

### Data Aggregation

Common functions used in data aggregation include:

Maximum (max): Finds the highest value in a group.
Mean (mean): Calculates the average value of a group.
Sum (sum): Calculates the total sum of a group.
Standard Deviation (std): Calculates the dispersion or spread of a group.

```python
scores=[89,76,92,54,88]
print("Maximum Score",max(scores))
print("Average Score",sum(scores)/len(scores))
```

### Aggregation in Pandas

```python
import pandas as pd
data= {
    'store': ['Store A', 'Store A', 'Store B', 'Store B', 'Store C'],
    'product': ['Apples', 'Bananas', 'Apples', 'Bananas', 'Apples'],
    'units_sold': [30, 50, 40, 35, 90],
    'price': [1.20, 0.50, 1.00, 0.50, 1.30]
}
df = pd.DataFrame(data)
print(df)
```

# Let’s find the maximum units sold and the average price of products by store using Pandas

```python
agg_func={'units_sold':'max','price':'mean'}
result = df.groupby('store').agg(agg_func)
print(result)
```

```python
import pandas as pd

data = {
    'store': ['Store A', 'Store A', 'Store B', 'Store B', 'Store C'],
    'product': ['Oranges', 'Mangoes', 'Oranges', 'Mangoes', 'Oranges'],
    'units_sold': [45, 60, 55, 45, 100],
    'price': [1.10, 1.50, 1.30, 1.60, 1.40]
}

df = pd.DataFrame(data)

# Calculate revenue
df['revenue'] = df['units_sold'] * df['price']

# Aggregate data to find the maximum units sold and total revenue by store
agg_funcs = {'units_sold': 'max', 'revenue': 'sum'}

result = df.groupby('store').agg(agg_funcs)
print(result)
```

### Understanding Correlation

Positive Correlation: When one variable increases, the other tends to increase. For example, studying more hours can lead to higher exam scores.

Negative Correlation: When one variable increases, the other tends to decrease. For instance, more TV time usually means less study time.

```python
import pandas as pd

# Sample dataset creation
data = {
    'Price': [300000, 450000, 200000, 350000, 500000],
    'Size': [1500, 2000, 1000, 1700, 2200],
    'Bedrooms': [3, 4, 2, 3, 4],
    'Age': [20, 15, 40, 10, 5]
}

# Create DataFrame
houses = pd.DataFrame(data)

# Display the DataFrame
print(houses)
```

### Correlation Calculation

```python
correlation_matirx=houses.corr()
print(correlation_matirx)
```

### Finding Correlation Between Two Columns

```python
correlation_price_size=houses['Price'].corr(houses['Size'])
print(correlation_price_size)
```
