# Question 67 — Customize Box Plot Outliers

### Task

Create a box plot comparing `total_bill` between smokers and non-smokers, and customize the appearance of the outlier points.

### Requirements

- [ ] Put `smoker` on the x-axis
- [ ] Put `total_bill` on the y-axis
- [ ] Customize the outlier marker using `flierprops`
- [ ] Make the outlier markers red
- [ ] Set the outlier marker size to 6
- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot


```python
flierprops = {
    "marker": "o",
    "markerfacecolor": "red",
    "markersize": 6
}

sns.boxplot(
    data=data,
    x="smoker",
    y="total_bill",
    flierprops=flierprops
)

plt.title("Total Bill by Smoking Status")
plt.xlabel("Smoker")
plt.ylabel("Total Bill")

plt.show()
```

# Question 69 — Multiple KDE Curves

### Task

Plot the distribution of `total_bill` separately for smokers and non-smokers using KDE curves.

### Requirements
- [ ] Use `total_bill` as the numerical variable
- [ ] Use `smoker` to create separate distributions
- [ ] Use `hue="smoker"`
- [ ] Add a title
- [ ] Add x-axis and y-axis labels

```python
sns.kdeplot(data=data,x="total_bill",hue="smoker")
plt.title("Total Bill Distribution by Smoking Status")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```
