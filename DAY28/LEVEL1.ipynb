# Question 56 — Seaborn Count Plot

### Task

Visualize how many customers visited on each day.

### Requirements

- [ ] Use `sns.countplot()`
- [ ] Put `day` on the x-axis
- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot

````python
sns.countplot(data=data,x='day')
plt.title("count plot")
plt.xlabel("Days")
plt.ylabel("customers")
plt.show()
````

# Question 57 — Seaborn Histogram

### Task

Visualize the distribution of `total_bill` using Seaborn.

### Requirements

- [ ] Use `sns.histplot()`
- [ ] Use `total_bill` on the x-axis
- [ ] Set `bins=10`
- [ ] Add a title
- [ ] Add an x-axis label
- [ ] Add a y-axis label
- [ ] Display the plot

````python
sns.histplot(data=data,x='total_bill',bins=10)
plt.title("total bill distribution")
plt.xlabel("total bill")
plt.ylabel("customers")
plt.show()
````

# Question 58 — Seaborn KDE Plot

### Task

Visualize the distribution of `total_bill` using a KDE plot.

````python
sns.kdeplot(data=data,x='total_bill')
````

````python
sns.rugplot(data=data,x='total_bill')
plt.title("Rug plot")
plt.xlabel("total bill")
plt.show()
````

# Question 60 — Combine Histogram and KDE

### Task

Create a visualization that shows both the distribution of `total_bill` and its smooth KDE curve.

### Requirements

- [ ] Use `sns.histplot()`
- [ ] Use `total_bill` on the x-axis
- [ ] Set `bins=10`
- [ ] Add the KDE curve using `kde=True`
- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot


````python
sns.histplot(data=data,x='total_bill',bins=10,kde=True)
plt.title("total bill with KDE")
plt.xlabel("Total bill")
plt.ylabel("count")
plt.show()
````
