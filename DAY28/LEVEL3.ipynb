# Question 71 — Scatter Plot with `hue`

### Task

Create a scatter plot showing the relationship between `total_bill` and `tip`, while using `smoker` to distinguish the observations.

### Requirements
- [ ] Put `total_bill` on the x-axis
- [ ] Put `tip` on the y-axis
- [ ] Use `smoker` as `hue`
- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot


```python
import matplotlib.pyplot as plt 
import seaborn as sns 
import pandas as pd
import numpy as np
```

```python
data=sns.load_dataset('tips')
sns.scatterplot(data=data,x='total_bill',y='tip',hue='smoker')
plt.title("smoker vs non smoker ")
plt.xlabel("total_bill")
plt.ylabel("tip")
plt.show()
```
