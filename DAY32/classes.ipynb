## creatinng a basic class

```python
class Cat:
    def __init__(self,name,age,owner):
        self.name=name
        self.age=age
        self.owner=owner
    def run(self):
        print(f"The {self.name} is running")
    def sleep(self):
        print(f"The {self.name} is not sleeping")
    def play(self):
        print(f"This is {self.owner} the owner of {self.name}  ")

nigga=Cat('umer',563,'Tanisha')

print("my cat name is ",nigga.name,"the Owner is ",nigga.owner)
nigga.sleep()
nigga.play()
```

```python
class Dog:
    def __init__(self,dogname,age):
        self.dogname=dogname 
        self.age=age 
    # behaviours
    def behave(self):
        print(f"my {self.dogname} should be posseive")


umer_dog= Dog("jafck",88)
print(f"umers dog name is {umer_dog.dogname} and he is possseive")
umer_dog.behave()
```

```python
class Bank:
    # members 
    def __init__(self,name,jobrole,salary):
        self.name=name
        self.jobrole=jobrole 
        self.salary=salary

    def salary_given(self):
        print(f" {self.name} salary is: {self.salary} and their job  is {self.jobrole} ")


umer_bhai= Bank("umer","manager",7403745024)
Nalla_bhai= Bank("Makoda","robber",15)

umer_bhai.salary_given()
Nalla_bhai.salary_given()

    
```
