# Regex Cheat Sheet for Pandas & AIML

## What is Regex?

Regex (Regular Expressions) is a pattern-matching language used to search, filter, and validate text.

Instead of searching for exact words, you search for **patterns**.

Example:

```python
df["Name"].str.contains("^A")
```

Finds all names starting with **A**.

---

# Most Important Regex Symbols

| Symbol | Meaning | Example | Matches |
|---------|---------|---------|---------|
| `^` | Starts with | `^A` | Alice, Alex |
| `$` | Ends with | `n$` | Helen, John |
| `|` | OR | `A|D` | A or D |
| `.` | Any single character | `A.` | Al, An, A1 |
| `.*` | Zero or more characters | `A.*` | Alice, Alex, Aman |

---

# Pandas Examples

## 1. Starts with A


# Handling Missing Values

If the column contains `NaN`:

```python
df["City"].str.contains("Del", na=False)
```

Always use:

```python
na=False
```

This treats missing values as `False` instead of throwing an error.

---

# When to Use What?

### Exact Match

```python
df["City"] == "Delhi"
```

---

### Partial Match

```python
df["City"].str.contains("Del")
```

---

### Starts With

```python
df["City"].str.startswith("Del")
```

---

### Ends With

```python
df["City"].str.endswith("hi")
```

---

### Regex Pattern

```python
df["Name"].str.contains("^A|^D")
```

# Regex You Need for AIML

Learn these first:

- `^` → Starts with
- `$` → Ends with
- `|` → OR
- `.` → Any character
- `.*` → Anything
- `\d` → Digit (0-9)
- `\w` → Letter, digit or underscore
- `\s` → Space

These cover around **80% of regex used in Pandas, data cleaning, and machine learning preprocessing.**

---

# Memory Box

| Task | Best Method |
|------|-------------|
| Exact match | `==` |
| Contains text | `str.contains()` |
| Starts with | `str.startswith()` |
| Ends with | `str.endswith()` |
| Regex search | `str.contains(regex)` |

---

# Interview Tip

Use the simplest method that solves the problem.

✅ Good:

```python
df["Name"].str.startswith("A")
```

❌ Overkill:

```python
df["Name"].str.contains("^A")
```

Regex is powerful, but only use it when pattern matching is actually required.
