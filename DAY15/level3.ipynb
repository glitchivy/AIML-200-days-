````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen"],
    "Age": [25, 30, np.nan, 35, 28, np.nan, 30, 26],
    "Department": ["HR", "IT", "IT", np.nan, "HR", "Finance", "IT", "Marketing"],
    "Salary": [45000, 60000, 55000, 80000, np.nan, 90000, 62000, 50000],
    "City": ["Mumbai", "Delhi", "Pune", "Mumbai", np.nan, "Pune", "Mumbai", "Delhi"]
})
````

# Pandas Level 3 - Question 45

---
