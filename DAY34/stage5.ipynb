```python
def mean_normalization(X):
    mean=np.mean(X,axis=0)
    maximum=np.max(X,axis=0)
    minimum= np.min(X,axis=0)
    X_scaled = (X-mean)/(maximum-minimum)
    return X_scaled

def z_score_normalization(X):
    mean=np.mean(X,axis=0)
    standard= np.std(X,axis=0)
    X_scaled = (X-mean)/standard
    
    # Your implementation
    return X_scaled

def simple_rescaling(X):
    X_scaled = X.copy()
    X_scaled[:, 0] = X_scaled[:, 0] / 1000
    # Divide first column by 1000
    # Keep other columns unchanged
    return X_scaled
```

```python
def fit_scaler(X_train):
    mean = np.mean(X_train, axis=0)
    std = np.std(X_train, axis=0)

    # Prevent division by zero
    safe_std = np.where(std == 0, 1, std)

    statistics = {
        "mean": mean,
        "std": safe_std
    }

    return statistics


def transform_scaler(X, statistics):
    mean = statistics["mean"]
    std = statistics["std"]

    X_scaled = (X - mean) / std

    return X_scaled
```

```python
def gradient_descent_converge(
    X, y, w, b,
    alpha,
    max_iterations=10000,
    tolerance=1e-8,
    patience=5
):
    cost_history = []
    stable_count = 0

    previous_cost = compute_cost(X, y, w, b)

    for i in range(max_iterations):

        dj_dw, dj_db = compute_gradient(X, y, w, b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        current_cost = compute_cost(X, y, w, b)
        cost_history.append(current_cost)

        improvement = abs(previous_cost - current_cost)

        if improvement < tolerance:
            stable_count += 1
        else:
            stable_count = 0

        if stable_count >= patience:
            print("Converged at iteration:", i + 1)
            break

        previous_cost = current_cost

    return w, b, cost_history
```

# feature enginerring with numpy 

```python
import numpy as np

X = np.array([
    [1000, 2, 5],
    [1500, 3, 10],
    [2000, 4, 3],
    [2500, 4, 8]
], dtype=float)
```

```python
area=X[:,0]
bedrooms=X[:,1]
age=X[:,2]

area_per_bedroom = area / bedrooms
area_age_interaction = area * age
area_squared = area ** 2
```

```python
X_engineered = np.column_stack((
    X,
    area_per_bedroom,
    area_age_interaction,
    area_squared
))

print(X_engineered)
print(X_engineered.shape)
```

## polynomail feature generation

```python
def polynomial_features(X,degree):
    features=[]

    for power in range(1,degree+1):
        features.append(X**power)
    return np.column_stack(features)

X = np.array([1, 2, 3, 4, 5])

X_poly = polynomial_features(X, degree=3)

print(X_poly)
```

```python
def prediction(X, w, b):
    return X @ w + b


def compute_cost(X, y, w, b):
    m = X.shape[0]
    error = prediction(X, w, b) - y
    return np.sum(error**2) / (2*m)


def compute_gradient(X, y, w, b):
    m = X.shape[0]
    error = prediction(X, w, b) - y

    dj_dw = (X.T @ error) / m
    dj_db = np.sum(error) / m

    return dj_dw, dj_db


def gradient_descent(X, y, w, b, alpha, iterations):
    cost_history = []

    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(X, y, w, b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        cost_history.append(compute_cost(X, y, w, b))

    return w, b, cost_history
```
