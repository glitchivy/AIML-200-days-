# FINAL REGRESSION PROJECT 

Phase	Work
1. Data preparation	Load, inspect and clean the dataset
2. Exploratory analysis	Understand features, distributions and relationships
3. Feature engineering	Create meaningful new features
4. Data splitting	Create training, validation and test sets
5. Polynomial features	Generate polynomial features
6. Feature scaling	Fit and apply scaling correctly
7. Model implementation	Implement prediction, cost and gradients
8. Gradient descent	Train your regression model
9. Optimization	Experiment with learning rates and convergence
10. Model evaluation	Calculate MAE, MSE and RMSE
11. Debugging	Perform gradient checks and investigate errors
12. Model selection	Compare linear and polynomial regression
13. Final testing	Evaluate on previously unseen data
