# IMPLEMENTING GRADIENT DESECEENT 

````python
# step 1 : implement the prediction function 
import numpy as np 

def prediction(X,w,b):
    return X@w + b 

# step 2 calculate error vector 

def errors(X,w,b,y):
    y_predicted= prediction(X,w,b)
    return y_predicted-y 

def compute_gradient(X,w,b,y):
    m=X.shape[0]
    error=errors(X,w,b,y)

    dj_dw= (X.T@error)/m 

    dj_db= np.sum(error)/m 

    return dj_dw,dj_db
````

````python
X = np.array([
    [1, 2],
    [2, 3],
    [3, 4]
])

y = np.array([10, 15, 20])

w = np.array([2, 3])

b = 1

dj_dw, dj_db = compute_gradient(X, w, b, y)

print("Weight gradient:", dj_dw)
print("Bias gradient:", dj_db)
print("Gradient shape:", dj_dw.shape)
````
