# Data Cleaning and Preprocessing

### Conditional Selection

```python
import pandas as pd 
data = {'Name': ['Alice', 'Bob', 'Charlie', 'Dave'], 'Score': [88, 92, 95, 80]}
scores_df = pd.DataFrame(data)
```

```python
print(scores_df[scores_df['Score']>80])
```

### Compound Conditional Selection

```python
print(scores_df[(scores_df['Score']>80) & (scores_df['Name'].str.startswith("A"))])
```

### what if we want all students except for 'Bob'. 

```python
print(scores_df[~(scores_df['Name']=='Bob')])
```

### where()

```python
print(scores_df['Score'].where(scores_df['Score']>85,other="Fail"))
```

```python
import pandas as pd

# DataFrame representing students and their test scores in two subjects
df = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Dana', 'Eli'],
    'math_score': [82, 95, 78, 98, 89],
    'history_score': [75, 88, 90, 64, 80]
})

# Select students who scored more than 85 in math or history
top_students = df[(df['math_score'] > 85) | (df['history_score'] > 85)]
print(top_students)
```

```python
import pandas as pd

# Grades DataFrame
grades = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'math_score': [91, 85, 79, 92, 88],
    'study_hours': [15, 8, 5, 18, 11]
})

# TODO: Select students with a math score <= 80 OR study_hours > 10
top_students = grades[(grades['math_score'] <= 80) | (grades['study_hours'] > 10)]
print(top_students)
```

```python
import pandas as pd

# DataFrame with students’ data
students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Daisy', 'Eve'],
    'grade': [85, 73, 95, 88, 76],
    'study_hours': [15.5, 9.0, 10.5, 12.0, 8.5]
})

# TODO: Write code to select only the students who have a grade above 80 and study more than 10 hours
print(students[(students['grade']>=80)& (students['study_hours']>10)])
```

### Handling Missing Values 

### Missing Data in Datasets

```python
import pandas as pd

data = {'Name': ['Anna', 'Bob', 'Charlie', 'David', None],
        'Score': [85, 88, None, 92, 90]}
df = pd.DataFrame(data)
print(df)
```

### Identifying Missing Values with Pandas

```python
print(df.isnull())
```

```python
print(df.notnull())
```

### Handling Missing Values: Removal

```python
print(df.dropna()) # method 1
```

```python
print(df.dropna(subset=['Score'])) #method 2
```

```python
print(df.fillna(0)) #mthod 3 filla missing values with zero
```

### forward fill and backward fill

```python
print(df.ffill())
```

```python
print(df.bfill())
```

### Handling Missing Values in Real-world Scenarios

```python
mean_score=df['Score'].mean()
print(df.fillna(mean_score))
```

```python
import pandas as pd

# Create dataset
grades = pd.DataFrame({'Student': ['Alice', 'Bob', 'Eve', 'Jack'], 'Math': [92, None, 87, 78], 'English': [None, 85, 88, 90]})

# filling the missing math grades with the mean math grade
grades['Math'] = grades['Math'].fillna(grades['Math'].mean())
print(grades)
```
