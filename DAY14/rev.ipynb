### Conditional Selection

```python
import pandas as pd 
data = {'Name': ['Alice', 'Bob', 'Charlie', 'Dave'], 'Score': [88, 92, 95, 80]}
scores_df = pd.DataFrame(data)
```

```python
print(scores_df[scores_df['Score']>80])
```

### Compound Conditional Selection

```python
print(scores_df[(scores_df['Score']>80) & (scores_df['Name'].str.startswith("A"))])
```

### what if we want all students except for 'Bob'. 

```python
print(scores_df[~(scores_df['Name']=='Bob')])
```

### where()

```python
print(scores_df['Score'].where(scores_df['Score']>85,other="Fail"))
```

```python
import pandas as pd
import numpy as np

# Grocery Store inventory with item names and their stock count
inventory = pd.DataFrame({"Item": ["Tomatoes", "Lettuce", "Potatoes"], "Stock": [25, 15, 40]})
# Create a new column "Reorder" where the value is "Yes" if Stock is less than 20, otherwise "No"
inventory["Reorder"] = np.where(inventory["Stock"] < 20, "Yes", "No")
print(inventory)
```

```python
import pandas as pd

# DataFrame for fruits inventory with price per unit and available stock
inventory = pd.DataFrame({'Fruit': ['Apples', 'Bananas', 'Grapes'], 'Price per Unit': [0.50, 0.20, 2.00], 'Stock': [100, 200, 80]})
# Create new column "Stock Value" which is Price per Unit * Stock
inventory['Stock Value'] = inventory['Price per Unit'] * inventory['Stock']
print(inventory)
```

```python
import pandas as pd
import numpy as np

# DataFrame with items, in-stock quantity and minimum required stock
inventory = pd.DataFrame({
    "Item": ["Tomatoes", "Potatoes", "Onions"],
    "In_Stock": [50, 20, 15],
    "Min_Required": [30, 15, 25]
})
# TODO: Check if in-stock quantity is lower than the minimum required stock and set 'Order' column to True or False accordingly
inventory["Order"] = np.where(inventory["In_Stock"]>inventory["Min_Required"],False,True)

print(inventory)
```

### Dealing with Inconsistencies in Data

```python
df.replace(["xl",'X-L'],'XL',inplace=True)
print(df.value_counts())
```

### Detecting and Filtering Outliers

```python
# A dataset with an outlier
data = [1, 1.2, 1.1, 1.05, 1.5, 1.4, 9]
df = pd.DataFrame(data, columns=['Values'])
#calculate q1 and q3 
q1=df['Values'].quantile(0.25)
q3=df['Values'].quantile(0.75)

#IQR
IQR=q3-q1

#range

L_B=q1-1.5*IQR
U_B=q3+1.5*IQR 


no_outliners=df[(df['Values']>L_B) & (df['Values']<U_B)]
print(no_outliners)

```

### Data Transformation

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# creating a Data frame

data = {'Feature1': [0.5, 0.6, 0.9], 'Feature2': [10, 12, 15]}
df = pd.DataFrame(data)


scaler = StandardScaler()

df['Feature2_scaled'] = scaler.fit_transform(df[['Feature2']])

print("\nOriginal Data with Scaled Feature2 Column:")
print(df)
```

```python
import pandas as pd

# Dataset of t-shirt sizes with inconsistencies
sizes = ['small', 'Large', 'medium', 'sm', 'lg', 'S', 'l', 'M']
df = pd.DataFrame(sizes, columns=['TShirtSize'])

# Standardize sizes using a dictionary with replacements
size_replacements = {'sm': 'small', 'S': 'small', 'lg': 'large', 'Large': 'large', 'l': 'large', 'M': 'medium'}
df['TShirtSize'] = df['TShirtSize'].replace(size_replacements)
print(df['TShirtSize'].value_counts())
```
