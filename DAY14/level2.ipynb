# Pandas Level 2 - Question 28

## Topic
**Select Rows Using Multiple Conditions**

---

## Problem Statement

Using the existing DataFrame (`df`), select all employees who satisfy **both** of the following conditions:

- `Age > 25`
- `Salary > 50000`

Store the result in a variable named `result` and print it.


```python
result=df[(df['Age']>25)&(df['Salary']>50000)]
print(result)
```

# Pandas Level 2 - Question 29

## Topic
**Select Rows Using `isin()`**

---

## Problem Statement

Using the existing DataFrame (`df`), select all employees who belong to either:

- `HR`
- `Finance`

Store the result in a variable named `result` and print it.


```python

result = df[df["Department"].isin(["HR", "Finance"])]
print(result)
```

# Pandas Level 2 - Question 30

## Topic
**Select Rows Using `between()`**

---

## Problem Statement

Using the existing DataFrame (`df`), select all employees whose **Age is between 25 and 35 (inclusive)**.

Store the result in a variable named `result` and print it.

```python
result=df[(df['Age']>24) & (df['Age']<36)]
print(result)

# other mehotd 
result = df[df["Age"].between(25, 35)]
print(result)
```
