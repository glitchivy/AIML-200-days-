# Pandas Level 1 - Question 1

## Topic
**Creating a Series from a Python List**

---

## Problem Statement

You are given the following Python list:

```python
[10, 20, 30, 40, 50]
```

Create a **Pandas Series** from this list.

````python
import pandas as pd
arr=[10,20,30,40,50]
series=pd.Series(arr)
print(series)
````

# Pandas Level 1 - Question 2

## Topic

---

## Problem Statement

You are given the following data:

```python
[95, 88, 76]
```

Assign the following custom index labels to the data:

```python
["Math", "Science", "English"]
```

Create a **Pandas Series** using the given data and custom index labels.

---


````python
import pandas as pd
arr=[95, 88, 76]
series=pd.Series(arr,index=["Math","Science","English"])
print(series)
````

# Pandas Level 1 - Question 3

## Topic
**Creating a DataFrame from a Dictionary**

---

## Problem Statement

You are given the following data:

- **Name:** Alice, Bob, Charlie
- **Age:** 25, 30, 35
- **City:** Mumbai, Delhi, Pune

Create a **Pandas DataFrame** using a Python dictionary.

````python
import pandas as pd
values={"Name": ["Alice", "Bob", "Charlie"],
 "Age": [25, 30, 35],
"City":[" Mumbai", "Delhi", "Pune"]}

df=pd.DataFrame(values)
print(df)
````

# Biggest Lessons

1.  Always identify whether you're working with a **Series** or a
    **DataFrame**.
2.  Know the difference between **methods** (`()`) and **attributes**
    (no `()`).
3.  Use the correct terminology:
    -   Rows vs Values
    -   Index vs Columns
    -   `dtype` vs `dtypes`
4.  If using a notebook, **Run All** after changing earlier cells to
    avoid stale state.
5.  Read the question carefully before coding.
