# Titanic Dataset: Feature Classification

## Feature Classification Table

|  # | Feature       | Meaning                                | Classification                    | Important Note                                  |
| -: | ------------- | -------------------------------------- | --------------------------------- | ----------------------------------------------- |
|  1 | `survived`    | Whether the passenger survived         | **Binary + Categorical + Target** | `0` = No, `1` = Yes                             |
|  2 | `pclass`      | Passenger ticket class                 | **Ordinal**                       | 1st > 2nd > 3rd in class ranking                |
|  3 | `sex`         | Passenger sex                          | **Categorical + Binary**          | `male` / `female`                               |
|  4 | `age`         | Passenger age                          | **Numerical**                     | Continuous numerical variable                   |
|  5 | `sibsp`       | Number of siblings/spouses aboard      | **Discrete Numerical**            | Counts people, so values are whole numbers      |
|  6 | `parch`       | Number of parents/children aboard      | **Discrete Numerical**            | Counts people, so values are whole numbers      |
|  7 | `fare`        | Passenger ticket fare                  | **Numerical**                     | Continuous numerical variable                   |
|  8 | `embarked`    | Port where passenger boarded           | **Categorical**                   | `C`, `Q`, `S`                                   |
|  9 | `class`       | Passenger class in text form           | **Ordinal**                       | First, Second, Third                            |
| 10 | `who`         | Passenger category                     | **Categorical**                   | `man`, `woman`, `child`                         |
| 11 | `adult_male`  | Whether passenger was an adult male    | **Boolean + Binary**              | `True` / `False`                                |
| 12 | `deck`        | Passenger's deck                       | **Categorical**                   | A, B, C, etc.; contains many missing values     |
| 13 | `embark_town` | Name of boarding town                  | **Categorical**                   | Southampton, Cherbourg, Queenstown              |
| 14 | `alive`       | Whether passenger was alive            | **Binary + Categorical**          | `yes` / `no`; essentially duplicates `survived` |
| 15 | `alone`       | Whether passenger was travelling alone | **Boolean + Binary**              | `True` / `False`                                |
