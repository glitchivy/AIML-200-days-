## Problem Statement

Using the given messy DataFrame (`df`), build a complete data cleaning pipeline.

Store the final cleaned DataFrame in a variable named `result` and print it.

## Tasks

Perform the following operations in order:

### 1.

Remove duplicate rows.

---

### 2.

Remove leading and trailing spaces from the **Name** column.

### 3.

Convert the **Joining_Date** column to datetime.

### 4.

Fill missing values in **Name** with:

```text

```

---

### 5.

Fill missing values in **Age** using the median.

---

### 6.

Fill missing values in **Salary** using the median.

### 7.

````python
df= df.drop_duplicates() #task1
print(df)
````

````python
df['Name'] = df['Name'].str.strip() #task2
print(df)
````

````python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])
print(df) #task3
````

````python
df["Name"]= df['Name'].fillna("Unknown") #task4
print(df)
````

````python
value=df["Age"].median()
df['Age']=df["Age"].fillna(value)
print(df)  #task5
````

````python
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
print(df) #task6
````

````python
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
print(df) #task7
````

````python
df=df[df["Age"]<=100]
print(df)
#task8
````

````python
df = df.reset_index(drop=True)

print(df)
````
