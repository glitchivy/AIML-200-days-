# Exploring Marker Customization

```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

Here are some common marker styles you can choose from:

Circle: 'o' - Represents each point as a circle.

Square: 's' - Uses squares for each data point.

Triangle: '^' - Displays data points as upward-pointing triangles.

Star: '*' - Marks each point with a star shape.

Diamond: 'D' - Uses diamond shapes to represent data points.

```python
penguins=sns.load_dataset('penguins')
plt.scatter(penguins['bill_length_mm'],penguins['body_mass_g'],marker='*',color='red')
plt.show()
```

### Multi-figure Layouts

Replace n_rows and n_columns with numbers to set up your grid size, like plt.subplots(2, 3) for two rows and three columns.

The function returns two objects:

fig: Represents the entire figure containing all plots.

ax: An array that holds each individual subplot.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Penguins dataset
penguins = sns.load_dataset('penguins')

# Create a horizontal layout with two subplots
fig, ax = plt.subplots(1, 2, figsize=(8, 4))

# First subplot: Histogram of bill lengths
ax[0].hist(penguins['bill_length_mm'], color='blue')
ax[0].set_title('Bill Length Distribution')
ax[0].set_xlabel('Bill Length (mm)')
ax[0].set_ylabel('Frequency')

# Second subplot: Scatter plot of flipper length vs. body mass
ax[1].scatter(penguins['flipper_length_mm'], penguins['body_mass_g'], color='red')
ax[1].set_title('Flipper Length vs Body Mass')
ax[1].set_xlabel('Flipper Length (mm)')
ax[1].set_ylabel('Body Mass (g)')

# Adjust and display layout
plt.tight_layout()
plt.show()
```

```python
# Create a vertical layout with two subplots
fig, ax = plt.subplots(2, 1, figsize=(4, 6))

# First subplot: Histogram of bill lengths
ax[0].hist(penguins['bill_length_mm'], color='blue')
ax[0].set_title('Bill Length Distribution')
ax[0].set_xlabel('Bill Length (mm)')
ax[0].set_ylabel('Frequency')

# Second subplot: Scatter plot of flipper length vs. body mass
ax[1].scatter(penguins['flipper_length_mm'], penguins['body_mass_g'], color='red')
ax[1].set_title('Flipper Length vs Body Mass')
ax[1].set_xlabel('Flipper Length (mm)')
ax[1].set_ylabel('Body Mass (g)')

# Adjust and display layout
plt.tight_layout()
plt.show()
```

### Bi-Dimensional Layout

just think of it like a matirx

```python
# Create a 2x2 grid layout to fit three plots
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

# First subplot: Histogram of bill lengths
ax[0, 0].hist(penguins['bill_length_mm'], color='blue')
ax[0, 0].set_title('Bill Length Distribution')
ax[0, 0].set_xlabel('Bill Length (mm)')
ax[0, 0].set_ylabel('Frequency')

# Second subplot: Scatter plot of flipper length vs. body mass
ax[0, 1].scatter(penguins['flipper_length_mm'], penguins['body_mass_g'], color='red')
ax[0, 1].set_title('Flipper Length vs Body Mass')
ax[0, 1].set_xlabel('Flipper Length (mm)')
ax[0, 1].set_ylabel('Body Mass (g)')

# Third subplot: Line plot of flipper lengths
ax[1, 0].plot(penguins.index, penguins['flipper_length_mm'], color='green')
ax[1, 0].set_title('Flipper Lengths Over Index')
ax[1, 0].set_xlabel('Index')
ax[1, 0].set_ylabel('Flipper Length (mm)')

# Set empty plot (ax[1, 1]) to keep layout balanced
ax[1, 1].axis('off')

# Adjust and display layout
plt.tight_layout()
plt.show()
```

# Basic Histogram Creation with Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
penguins = sns.load_dataset('penguins')

# Create a basic histogram
sns.histplot(data=penguins, x='body_mass_g')

# Add title and labels
plt.title('Penguin Body Mass Distribution')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')

# Display the plot
plt.show()
```

```python
# Create a histogram with custom bin numbers
sns.histplot(data=penguins, x='body_mass_g', bins=30)
```
