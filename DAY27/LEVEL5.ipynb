# Question 41 — Annotate a Point

### Task

Create a line plot and annotate the highest sales value.

### Data

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]
max_sales=max(sales)
index=sales.index(max_sales)
plt.plot(months,sales)
plt.annotate("High_sales",xy=(months[index],max_sales))
plt.show()
````

# Question 42 — Horizontal Reference Line

### Task

Create a line plot of monthly sales and add a horizontal reference line showing the **sales target**.

### Data

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]

target = 40

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]

target = 40

plt.plot(months, sales, marker='o')

plt.axhline(y=target, label="Target")

plt.title("Monthly Sales vs Target")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.show()
````

# Question 43 — Vertical Reference Line

### Data

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20, 35, 30, 50, 45, 60]

target_month = "Apr"

index = months.index(target_month)

plt.plot(months, sales, marker='o')

plt.axvline(x=index, label="Target Month")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.show()
````

# Question 44 — Fill Between Two Lines

### Task

Create a chart comparing actual sales with target sales and shade the area where actual sales are above the target.

### Data

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

actual_sales = [20, 35, 30, 50, 45, 60]
target_sales = [25, 25, 25, 40, 40, 40]

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

actual_sales = [20, 35, 30, 50, 45, 60]
target_sales = [25, 25, 25, 40, 40, 40]

above_target = np.array(actual_sales) > np.array(target_sales)

plt.plot(months, actual_sales, label="Actual Sales")
plt.plot(months, target_sales, label="Target Sales")

plt.fill_between(
    months,
    actual_sales,
    target_sales,
    where=above_target,
    alpha=0.3,
    label="Above Target"
)

plt.title("Actual Sales vs Target")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
````
