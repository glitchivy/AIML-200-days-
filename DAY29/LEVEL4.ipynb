```python

import pandas as pd

data = pd.DataFrame({
    "department": [
        "HR", "HR", "HR", "HR", "HR",
        "IT", "IT", "IT", "IT", "IT",
        "Sales", "Sales", "Sales", "Sales", "Sales",
        "Finance", "Finance", "Finance", "Finance", "Finance",
        "IT", "Sales", "HR", "Finance", "IT",
        "Sales", "HR", "Finance", "IT", "Sales"
    ],

    "gender": [
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Female", "Male", "Female", "Male", "Female"
    ],

    "education": [
        "Bachelor", "Master", "Bachelor", "Master", "PhD",
        "Bachelor", "Master", "PhD", "Master", "Bachelor",
        "Bachelor", "Master", "Bachelor", "PhD", "Master",
        "Master", "Bachelor", "PhD", "Master", "Bachelor",
        "PhD", "Bachelor", "Master", "PhD", "Bachelor",
        "Master", "Bachelor", "Master", "PhD", "Bachelor"
    ],

    "salary": [
        42000, 48000, 45000, 52000, 60000,
        65000, 72000, 90000, 78000, 68000,
        45000, 52000, 49000, 65000, 58000,
        62000, 70000, 85000, 76000, 68000,
        82000, 55000, 47000, 80000, 74000,
        60000, 50000, 72000, 88000, 57000
    ]
})

data.head()
```

# Question 81 — Average Value by Category

### Task

Using a categorical dataset containing `department` and `salary`, create a bar plot showing the average salary for each department.

### Requirements

- [ ] Use `sns.barplot()`
- [ ] Put `department` on the x-axis
- [ ] Put `salary` on the y-axis
- [ ] Show the average salary for each department
- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot


```python
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np 
```

```python
sns.barplot(data=data,x='department',y='salary',estimator="mean")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.show()
```
