```python
import seaborn as sns 
import matplotlib.pyplot as plt 
import pandas as pd
```

# Question 75 — Joint Plot

### Task

Create a joint plot to visualize the relationship between `total_bill` and `tip`.

### Requirements

- [ ] Put `total_bill` on the x-axis
- [ ] Put `tip` on the y-axis
- [ ] Use the default plot kind
- [ ] Display the plot


```python
data=sns.load_dataset('tips')
sns.jointplot(data=data,x='total_bill',y='tip')
```

# Question 78 — Residual Plot

### Task

Create a residual plot to analyze the relationship between `total_bill` and `tip`.

### Requirements

- [ ] Add a title
- [ ] Add x-axis and y-axis labels
- [ ] Display the plot


```python
sns.residplot(data=data,x='total_bill',y='tip')
plt.xlabel("total bill")
plt.ylabel("tip")
plt.show()
```

Create an LM plot to visualize the relationship between `total_bill` and `tip`, including the regression line.

### Requirements

- [ ] Use `sns.lmplot()`
- [ ] Put `total_bill` on the x-axis
- [ ] Put `tip` on the y-axis
- [ ] Add a title
- [ ] Display the plot


```python

sns.lmplot(
    data=data,
    x="total_bill",
    y="tip")
plt.xlabel('total_bill')
plt.ylabel("tip")
plt.title("LM PLOT")
plt.show()
```

# Question 80 — Facet Grid

### Task

Create a facet grid to compare the relationship between `total_bill` and `tip` separately for each `time` category.

### Requirements

- [ ] Use `data` as the dataset
- [ ] Use `time` to create separate columns
- [ ] Map a scatter plot onto the grid
- [ ] Add a title
- [ ] Display the plot

```python
g = sns.FacetGrid(data, col="time")
g.map(sns.scatterplot, "total_bill", "tip")
g.fig.suptitle("Total Bill vs Tip by Time")
plt.tight_layout()
```
