# Question 94 — Upper Triangle Correlation Heatmap

### Task

Create a correlation heatmap that displays only the lower triangle of the correlation matrix.

### Requirements

- [ ] Calculate the correlation matrix
- [ ] Create a mask using `np.triu()`
- [ ] Pass the mask to `sns.heatmap()`
- [ ] Show correlation values
- [ ] Display the plot

```python
import numpy as np

corr = data.select_dtypes(include="number").corr()

mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f"
)

plt.show()
```

# Question 100 — Choose the Right Visualization

### Task

You are given the following business questions.

For each question, choose the most appropriate Seaborn visualization and create it.

### Question 1

Which department has the highest average salary?

### Question 2

How is salary distributed within each department?

### Question 3

Is there a relationship between experience and salary?

### Question 4

Does the experience-salary relationship differ between departments?

### Question 5

Which department has the most employees?

### Requirements

Choose an appropriate Seaborn plot for each question.

You should use different visualizations where appropriate.

For each visualization:

- [ ] Write the Seaborn code
- [ ] Add a meaningful title
- [ ] Add appropriate axis labels
- [ ] Display the plot



```python
# Q1 — Highest Average Salary by Department

sns.barplot(
    data=data,
    x="department",
    y="salary"
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()


# Q2 — Salary Distribution by Department

sns.boxplot(
    data=data,
    x="department",
    y="salary"
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()


# Q3 — Experience vs Salary

sns.scatterplot(
    data=data,
    x="experience",
    y="salary"
)

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()


# Q4 — Experience vs Salary by Department

sns.scatterplot(
    data=data,
    x="experience",
    y="salary",
    hue="department"
)

plt.title("Experience vs Salary by Department")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()


# Q5 — Number of Employees by Department

sns.countplot(
    data=data,
    x="department"
)

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```
